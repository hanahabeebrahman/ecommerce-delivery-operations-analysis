```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    r"C:\Users\ASUS\OneDrive\Desktop\ecommerce-delivery-operations-analysis/data/raw/DataCoSupplyChainDataset.csv",
    encoding="latin1"
)

print("Dataset loaded successfully!")
```

```python
clean_df = df.copy()

print("Original shape:", df.shape)
print("Working copy shape:", clean_df.shape)
```

```python
# Remove irrelevant and unnecessary columns

columns_to_drop = [
    "Customer Email",
    "Customer Fname",
    "Customer Lname",
    "Customer Password",
    "Customer Street",
    "Customer Zipcode",
    "Latitude",
    "Longitude",
    "Order Zipcode",
    "Product Description",
    "Product Image"
]

clean_df.drop(columns=columns_to_drop, inplace=True)

print("Shape after dropping columns:", clean_df.shape)
```

```python
# Convert date columns

date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for col in date_columns:
    clean_df[col] = pd.to_datetime(
        clean_df[col],
        errors="coerce"
    )

clean_df[date_columns].dtypes
```

```python
missing_values = clean_df.isnull().sum()

missing_values[missing_values > 0].sort_values(
    ascending=False
)
```

```python
print(
    "Invalid order dates:",
    clean_df["order date (DateOrders)"].isna().sum()
)

print(
    "Invalid shipping dates:",
    clean_df["shipping date (DateOrders)"].isna().sum()
)
```

```python
# Validate shipping durations

print(
    "Records with negative shipping days:",
    (clean_df["Days for shipping (real)"] < 0).sum()
)

print(
    "Records with negative scheduled days:",
    (clean_df["Days for shipment (scheduled)"] < 0).sum()
)

print(
    "Records where shipping date precedes order date:",
    (
        clean_df["shipping date (DateOrders)"]
        < clean_df["order date (DateOrders)"]
    ).sum()
)
```

```python
import os

os.makedirs(r"C:\Users\ASUS\OneDrive\Desktop\ecommerce-delivery-operations-analysis/data/cleaned", exist_ok=True)

clean_df.to_csv(
    r"C:\Users\ASUS\OneDrive\Desktop\ecommerce-delivery-operations-analysis/data/cleaned/supply_chain_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
print("Final shape:", clean_df.shape)
```
